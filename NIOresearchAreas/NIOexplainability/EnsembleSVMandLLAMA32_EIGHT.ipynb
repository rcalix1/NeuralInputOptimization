```python

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
import random

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, classification_report
```

```python

# ============================================================
# 1. Load CSV (unchanged)
# ============================================================

SEED = 42
EPOCHS = 200

np.random.seed(SEED)
torch.manual_seed(SEED)
```

```python

df = pd.read_csv("cyber_dataset_C.csv")
```

```python

df
```

```python

encoder = LabelEncoder()

y = encoder.fit_transform(df["tool_ground_truth"])
```

```python

num_classes = len(encoder.classes_)

num_classes 
```

```python

class_to_id = {
    name: i for i, name in enumerate(encoder.classes_)
}

class_to_id
```

```python

y_svm = np.array([
    class_to_id[name]
    for name in df["pred_svm"]
])



y_svm.shape



```

```python

y_svm
```

```python

y_llama = np.array([
    class_to_id.get(name, -1)
    for name in df["tool_predicted"]
])
```

```python

# Unknown is encoded as -1.

set ( [  name for name in df["tool_predicted"]   ]  )
```

```python

print(np.sum(y_llama == -1))
```

```python

# ============================================================
# 2. Train / validation / test split
# ============================================================

indices = np.arange(len(df))

indices
```

```python

random_seed = int( random.random() * 100 )     ## 42

## X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=random_seed)

trainval_idx, test_idx = train_test_split(
    indices,
    test_size=0.20,
    random_state=random_seed,             ## SEED,
    stratify=y
)
```

```python

print(  len(trainval_idx)  )
print(   len(test_idx   )  ) 
```

```python

test_idx
```

```python

train_idx = trainval_idx
```

```python

print("Training:", len(train_idx))
print("Testing:", len(test_idx))

```

```python

# ============================================================
# 3. Features: SVM + Llama predictions (23 features)
# ============================================================

# SVM has 11 classes
svm_onehot = np.zeros((len(df), num_classes))

svm_onehot.shape
```

```python

y_svm
```

```python

valid_svm = y_svm >= 0

print(  np.sum( valid_svm )  )
print(     len( valid_svm )  )
```

```python


svm_onehot[ 
    np.where(valid_svm)[0], y_svm[valid_svm] 
] = 1
```

```python

for thing in svm_onehot:
    print(  thing  )
```

```python

svm_onehot.shape
```

```python

# Llama has 11 classes + Unknown
llama_onehot = np.zeros((len(df), num_classes + 1))

llama_onehot.shape
```

```python

llama_idx = np.where(y_llama >= 0, y_llama, num_classes)
```

```python


print(     len(  llama_idx )  )
```

```python

'''
import numpy as np

y_llama_rc = np.array([2, -1, 4, 0, -1])
num_classes_rc = 11

llama_idx_rc = np.where(y_llama_rc >= 0, y_llama_rc, num_classes_rc)

print("Original:", y_llama_rc)
print("Modified:", llama_idx_rc)

'''
```

```python


llama_onehot[ 
      np.arange(len(df)), llama_idx
] = 1
```

```python

for thing in llama_onehot:
    print( thing )
```

```python


X = np.hstack((svm_onehot, llama_onehot)).astype(np.float32)

X.shape
```

```python

X_train = torch.tensor( X[train_idx] )
X_test  = torch.tensor( X[test_idx]  )

print("Input features:", X.shape[1])
```

```python

print(  X_train.shape  )
print(  X_test.shape   )
```

```python

# ============================================================
# 4. Neural network targets (11 cybersecurity tools)
# ============================================================

target = y

target.shape
```

```python


target_train = torch.tensor(
    target[train_idx],
    dtype=torch.long
)

```

```python

target_train.shape
```

```python

target_train
```

```python

# ============================================================
# 5. Neural network (11 cybersecurity tools)
# ============================================================

class SelectorNet(nn.Module):

    def __init__(self, input_dim):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(input_dim, 32),
            nn.ReLU(),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, 11)
        )

    def forward(self, x):
        return self.net(x)
```

```python



model = SelectorNet(  X.shape[1]  )


```

```python

X.shape[1]
```

```python

criterion = nn.CrossEntropyLoss()
```

```python

optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python



# ============================================================
# Train
# ============================================================

for epoch in range(EPOCHS):

    model.train()
    optimizer.zero_grad()

    logits = model(  X_train  )

    loss = criterion(logits, target_train)

    loss.backward()
    optimizer.step()

    if (epoch + 1) % 20 == 0:
        print(
            f"Epoch {epoch+1:3d} | "
            f"Loss = {loss.item():.4f}"
        )



```

```python

model.eval()

with torch.no_grad():
    ensemble_pred_res = torch.argmax(model(X_test), dim=1).numpy()
```

```python

ensemble_pred_res.shape
```

```python

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    confusion_matrix,
    classification_report
)

y_true = y[test_idx]

```

```python


print("Accuracy:", accuracy_score(y_true, ensemble_pred_res))

print("Macro F1:", f1_score(      y_true, ensemble_pred_res, average="macro" ))

print("Confusion Matrix:")
print(           confusion_matrix(y_true, ensemble_pred_res))

print("Classification Report:")
print(classification_report(      y_true, ensemble_pred_res,  zero_division=0   ))
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# The 14 cases where Llama returned Unknown
unknown_idx = np.flatnonzero(y_llama == -1)

# Actual tool labels for these cases
actual = y[unknown_idx]

# NIO optimized SVM features (first 11 columns)
svm_nio = X_nio.detach().numpy()[:, :11]

# Tool names
tools = list(encoder.classes_)

# Build 11 x 11 matrix
matrix = np.full((11, 11), np.nan)

for j in range(11):

    # Find samples whose actual tool is class j
    selected = actual == j

    if np.any(selected):

        # Average NIO SVM activations for that class
        matrix[:, j] = svm_nio[selected].mean(axis=0)

# Display labeled matrix
result = pd.DataFrame(
    matrix,
    index=tools,
    columns=tools
)

plt.figure(figsize=(10, 7))
plt.imshow(matrix, cmap="Blues", vmin=0, vmax=1)

plt.xticks(range(11), tools, rotation=90)
plt.yticks(range(11), tools)

plt.xlabel("Actual Tool")
plt.ylabel("NIO Optimized SVM Feature")
plt.title("NIO SVM Feature Associations — 14 Llama Unknown Cases")

plt.colorbar(label="Mean NIO activation")

plt.tight_layout()
plt.savefig("NIO_SVM_Associations.png", dpi=300, bbox_inches="tight")
plt.show()

result.round(2)
```
